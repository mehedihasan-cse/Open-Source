```python
import pandas as pd
from newspaper import Article
import time
```

```python
# text extraction symbol fix
def clean_text(text):
    if text:
        try:
            text = text.encode('latin1').decode('utf-8')
        except:
            pass
    return text
```

```python
# Load your dataset
df = pd.read_csv("GKG Dataset.csv")

```

```python
# Create empty columns
df['Extracted_Text'] = None
df['Word_Count'] = 0

# Function to extract article text
def extract_text(url):
    try:
        article = Article(url)
        article.download()
        article.parse()

        text = clean_text(article.text)
        word_count = len(text.split())

        return text, word_count

    except:
        return None, 0
```

```python
# Loop through URLs (IMPORTANT: limit first for testing)
for i, url in enumerate(df['DocumentIdentifier']):

    text, wc = extract_text(url)

    df.at[i, 'Extracted_Text'] = text
    df.at[i, 'Word_Count'] = wc

    print(f"{i+1}/{len(df)} done")

    # Small delay to avoid blocking
    time.sleep(1)
```
